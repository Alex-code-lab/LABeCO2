```python
import pandas as pd
tableau_masse_conso = pd.read_csv("masses_consommable_bis.csv")
# Remplace les virgules par des points dans la colonne
tableau_masse_conso['Masse unitaire (g)'] = (
    tableau_masse_conso['Masse unitaire (g)']
    .astype(str)  # S'assure que les données sont bien des chaînes
    .str.replace(',', '.', regex=False)  # Remplace les virgules par des points
    .astype(float)  # Convertit les chaînes en nombres flottants
)

# Vérifiez le résultat
tableau_masse_conso
```
